```python
ops = ['+', '-', '*']

def add(a, b):
    print("Result:", a + b)

def sub(a, b):
    print("Result:", a - b)

def mul(a, b):
    print("Result:", a * b)

def save_user():
    username = input("Create username: ")
    password = input("Create password: ")

    with open("users.txt", "a") as file: 
        file.write(username + "," + password + "\n")

    print("User saved successfully!\n")

def login():
    u = input("Enter username: ")
    p = input("Enter password: ")

    try:
        with open("users.txt", "r") as file:
            for line in file:
                user, pwd = line.strip().split(",")
                if u == user and p == pwd:
                    print("Login successful \n")
                    return True
    except FileNotFoundError:
        print("No users found. Please register first.\n")
        return False

    print("Invalid username/password \n")
    return False

def new_calc():
    print("Calculator")
    print("+ for addition")
    print("- for subtraction")
    print("* for multiplication")

    num1 = int(input("Enter num1: "))
    num2 = int(input("Enter num2: "))
    choice = input("Enter your op (+, -, *): ")

    if choice in ops:
        if choice == '+':
            add(num1, num2)
        elif choice == '-':
            sub(num1, num2)
        elif choice == '*':
            mul(num1, num2)
    else:
        print("Enter correct choice")

while True:
    print("\n1. Register")
    print("2. Login & Use Calculator")
    print("3. Exit")

    ch = input("Choose option: ")

    if ch == "1":
        save_user()

    elif ch == "2":
        if login():
            new_calc()

    elif ch == "3":
        print("Goodbye")
        break

    else:
        print("Invalid choice")
```
